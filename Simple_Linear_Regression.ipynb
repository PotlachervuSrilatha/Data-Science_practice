```python
# Step 1: Importing libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

```python
# Step 2: Prepare data
data = pd.DataFrame({
    'X': np.arange(1, 11),  # Independent variable
    'y': np.array([3, 4, 5, 6, 5, 7, 8, 9, 10, 11])  # Dependent variable
})
```

```python
X = data[['X']]  # Independent variable
y = data['y']    # Dependent variable
```

```python
# Step 3: Create a linear regression model object
model = LinearRegression()
```

```python
# Step 4: Fit the model
model.fit(X, y)
```

```python
# Step 5: Make predictions
y_pred = model.predict(X)
```

```python
# Step 5: Make predictions
y_pred = model.predict(X)
```

```python
# Step 6: Evaluate the model
mse = mean_squared_error(y, y_pred)

print(f'Mean Squared Error: {mse:.2f}')
```

```python
# Plotting the actual vs predicted values
plt.scatter(X, y, color='blue', label='Actual values')
plt.plot(X, y_pred, color='red', label='Predicted values')
plt.title('Actual vs Predicted values')
plt.xlabel('X')
plt.ylabel('y')
plt.legend()
plt.show()
```

```python
import warnings
warnings.filterwarnings('ignore')
# Example prediction for a new value
new_X = [[12]]
predicted_value = model.predict(new_X)
print(f'Predicted value for X={new_X[0][0]}: {predicted_value[0]:.2f}')
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Generating some example data
np.random.seed(0)
x = np.linspace(0, 10, 100)
y = 2 * x + 5 + np.random.normal(0, 2, 100)  # y = 2x + 5 + noise

# Creating the plot
plt.figure(figsize=(10, 6))

# Plotting the data points
plt.scatter(x, y, color='blue', label='Data points')

# Plotting the regression line
m = 2  # Slope
c = 5  # Intercept
plt.plot(x, m * x + c, color='red', label='Regression line: y = 2x + 5')

# Marking the intercept
plt.scatter(0, c, color='green', zorder=5)
plt.text(0.5, c + 1, 'Intercept (c) = 5', color='green')

# Marking the slope
plt.arrow(1, m * 1 + c, 1, m, head_width=0.5, head_length=0.5, fc='orange', ec='orange')
plt.text(2.5, m * 1 + c + 1, 'Slope (m) = 2', color='orange')

# Adding labels and legend
plt.xlabel('House Size (x)')
plt.ylabel('House Price (y)')
plt.title('Simple Linear Regression: y = mx + c')
plt.legend()
plt.grid(True)
plt.show()
```
